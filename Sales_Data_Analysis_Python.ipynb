```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Product": ["Laptop", "Phone", "Tablet", "Headphones", "Keyboard",
                "Laptop", "Phone", "Tablet", "Headphones", "Keyboard"],
    "Sales": [50000, 30000, 20000, 10000, 5000,
              45000, 35000, 18000, 12000, 7000],
    "Quantity": [2, 3, 4, 5, 6, 1, 4, 3, 6, 8]
}

df = pd.DataFrame(data)
df
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns)
print("\nData Types:")
print(df.dtypes)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
total_sales = df["Sales"].sum()
print("Total Sales:", total_sales)
```

```python
average_sales = df["Sales"].mean()
print("Average Sales:", average_sales)
```

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

print(product_sales)
```

```python
best_product = product_sales.idxmax()
best_sales = product_sales.max()

print("Best-Selling Product:", best_product)
print("Sales:", best_sales)
```

```python
product_sales.plot(kind="bar")

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
product_quantity = df.groupby("Product")["Quantity"].sum()

product_quantity.plot(kind="bar")

plt.title("Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Quantity")
plt.xticks(rotation=45)
plt.show()
```

```python
print("📊 SALES DATA ANALYSIS")
print("----------------------")
print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Best-Selling Product:", best_product)
print("Highest Product Sales:", best_sales)
```
